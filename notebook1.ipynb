```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load Data
gscpi = pd.read_excel("Data/gscpi_data.xls", sheet_name="Data", parse_dates=True)
wb_stress = pd.read_excel("Data/Stress index tables.xlsx", sheet_name="global", parse_dates=True)

# Inspect
print(gscpi.head())
print(wb_stress.head())
```

```python
gscpi.Date = pd.to_datetime(gscpi.Date)
```

```python
wb_stress.Departure_YearMonth = pd.to_datetime(wb_stress.Departure_YearMonth, format="%Y%m%d")
```

```python
gscpi.columns = ["Date", "GSCPI"]
wb_stress.columns = ["Date", "WB_Stress"]

data_nearest = pd.merge_asof(
    gscpi.sort_values("Date"),
    wb_stress.sort_values("Date"),
    on="Date",
    direction="nearest",
    tolerance=pd.Timedelta("30D"),
)
data_nearest.head()
```

```python
# Keep dates with observations from both indices.
data = data_nearest.dropna(subset=["GSCPI", "WB_Stress"]).copy()
```

```python
data.head()
```

```python
# The index levels use different units, so plot them on separate axes.
fig, axes = plt.subplots(2, 1, sharex=True, figsize=(10, 6))
axes[0].plot(data["Date"], data["GSCPI"])
axes[0].set_ylabel("NY Fed GSCPI")
axes[1].plot(data["Date"], data["WB_Stress"])
axes[1].set_ylabel("World Bank delayed capacity")
fig.tight_layout()
plt.show()

print(data[["GSCPI", "WB_Stress"]].describe())
data[["GSCPI", "WB_Stress"]].corr()
```
